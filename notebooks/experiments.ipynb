## 1. Model Initialization
Let's initialize the sub-models and the fusion model.

```python
image_model = ImageFeatureExtractor(feature_dim=512, pretrained=True)
video_model = VideoFeatureExtractor(feature_dim=512, num_frames=16, pretrained=True)
fusion_model = FusionModel(image_feature_dim=512, video_feature_dim=512, pretrained=True)

print(f"Fusion Model Trainable Params: {fusion_model.get_num_params()['trainable']:,}")
```

## 2. Forward Pass Test
Simulate an image (environment) and a video clip (motion).

```python
# Mock data: (B, C, H, W) and (B, T, C, H, W)
dummy_image = torch.randn(1, 3, 224, 224)
dummy_video = torch.randn(1, 16, 3, 224, 224)

with torch.no_grad():
    risk_score = fusion_model(dummy_image, dummy_video)
    assessment = fusion_model.get_risk_assessment(risk_score)

print(f"Risk Score: {risk_score.item():.4f}")
print(f"Status: {assessment['status']}")
print(f"Message: {assessment['message']}")
```
